
# NLTK API: Financial News Sentiment Analysis

## Description

This notebook introduces the Natural Language Toolkit (NLTK) and demonstrates its main tools for processing and analyzing financial news headlines.

We will explore tokenization, stopword removal, lemmatization, word frequency analysis, VADER sentiment scoring, and Naive Bayes classification.

The goal is to understand how NLTK can be used to classify financial news into positive, negative, and neutral sentiment.

## Imports

````python
%load_ext autoreload
%autoreload 2

# Standard libraries.
import logging
import re

# Data handling and visualization.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# NLTK tools.
import nltk
from nltk import FreqDist, NaiveBayesClassifier, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer
````

````python
# Download the NLTK resources required for this tutorial.
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("vader_lexicon")
````

````python
# Example financial news headline.
headline = "Apple shares rise 5% after strong quarterly earnings."

# Split the headline into tokens.
tokens = word_tokenize(headline)

print("Original headline:", headline)
print("Tokens:", tokens)
````

````python
# Load English stopwords.
stop_words = set(stopwords.words("english"))

# Remove stopwords and punctuation.
filtered_tokens = [
    word.lower()
    for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("Original tokens:", tokens)
print("Filtered tokens:", filtered_tokens)
````

````python
# Initialize the WordNet lemmatizer.
lemmatizer = WordNetLemmatizer()

# Convert words to their base forms.
lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

print("Before lemmatization:", filtered_tokens)
print("After lemmatization:", lemmatized_tokens)
````

````python
# Sample financial news headlines.
headlines = [
    "Apple shares rise after strong earnings",
    "Tesla shares fall after weak earnings",
    "Microsoft reports strong quarterly earnings",
    "Apple reports record quarterly profit",
    "Tesla shares rise after positive news"
]

# Tokenize and clean the headlines.
all_tokens = []

for text in headlines:
    words = word_tokenize(text.lower())
    cleaned_words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]
    all_tokens.extend(cleaned_words)

# Calculate word frequencies.
word_freq = FreqDist(all_tokens)

print("Top 10 most common words:")
print(word_freq.most_common(10))
````

````python
# Initialize the VADER sentiment analyzer.
sia = SentimentIntensityAnalyzer()

# Analyze sample financial news headlines.
sample_headlines = [
    "Company reports excellent profits and strong growth",
    "Company suffers heavy losses and declining revenue",
    "Company announces its quarterly earnings report"
]

for headline in sample_headlines:
    scores = sia.polarity_scores(headline)

    if scores["compound"] >= 0.05:
        sentiment = "Positive"
    elif scores["compound"] <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Headline:", headline)
    print("Sentiment:", sentiment)
    print("Scores:", scores)
    print("-" * 50)
````

````python
# Small labeled dataset for demonstrating Naive Bayes.
training_data = [
    ("Company reports strong profits", "positive"),
    ("Stock prices rise after excellent earnings", "positive"),
    ("Revenue growth exceeds expectations", "positive"),
    ("Company announces record profits", "positive"),
    ("Company reports heavy losses", "negative"),
    ("Stock prices fall after weak earnings", "negative"),
    ("Revenue declines below expectations", "negative"),
    ("Company faces serious financial problems", "negative"),
    ("Company announces quarterly results", "neutral"),
    ("Board meeting scheduled for Monday", "neutral"),
    ("Company releases annual report", "neutral"),
    ("Shareholders attend annual meeting", "neutral")
]

# Convert text into bag-of-words features.
def extract_features(text):
    words = word_tokenize(text.lower())
    return {
        word: True
        for word in words
        if word.isalpha()
    }

# Prepare the training examples.
train_features = [
    (extract_features(text), label)
    for text, label in training_data
]

# Train the NLTK Naive Bayes classifier.
classifier = NaiveBayesClassifier.train(train_features)

# Test on an example headline.
test_headline = "Company reports strong revenue growth"
prediction = classifier.classify(
    extract_features(test_headline)
)

print("Headline:", test_headline)
print("Predicted sentiment:", prediction)
````

````python
# Display the most informative features learned by Naive Bayes.
print("Top 10 most informative features:")
classifier.show_most_informative_features(10)
````
