```python
import sqlite3
import numpy as np
import pandas as pd
```

```python
db_path = "sqlite:///data.db"
```

```python
df = pd.DataFrame(
    {
        "date": pd.date_range("2024-01-01", periods=10),
        "sales": np.random.randint(50, 500, 10),
        "region": np.random.choice(["North", "South", "East", "West"], 10),
    }
)

df
```

```python
df.to_sql(
    "my_table",
    db_path,
    # if_exists="replace",
    if_exists="append",
    index=False,
)
```

```python
conn = sqlite3.connect("data.db")

df2 = pd.read_sql_query("SELECT * FROM my_table", conn)
print(df2)

conn.close()
```

```python
query = """
SELECT *
FROM my_table
WHERE date >= '2024-01-01'
AND sales > 100
"""

df3 = pd.read_sql_query(query, db_path)
df3
```
